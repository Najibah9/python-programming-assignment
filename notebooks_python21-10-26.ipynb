```python
def find_max_min(numbers):
    # Initialize max and min with the first element
    maximum = numbers[0]
    minimum = numbers[0]
    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    return maximum, minimum

def calculate_average(numbers):
    total = 0
    for num in numbers:
        total += num
    return total / len(numbers)

def print_even_odd(numbers):
    evens = [num for num in numbers if num % 2 == 0]
    odds = [num for num in numbers if num % 2 != 0]
    print(f"Even numbers: {evens}")
    print(f"Odd numbers:  {odds}")

# Main Program
numbers = []
print("Enter 10 integers:")
for i in range(10):
    num = int(input(f"Enter number {i + 1}: "))
    numbers.append(num)

max_val, min_val = find_max_min(numbers)
avg_val = calculate_average(numbers)

print(f"\nList: {numbers}")
print(f"Maximum value: {max_val}")
print(f"Minimum value: {min_val}")
print(f"Average value: {avg_val:.2f}")
print_even_odd(numbers)
```

```python
def reverse_array_manually(arr):
    n = len(arr)
    reversed_arr = []
    # Loop from the last index (n - 1) down to 0
    for i in range(n - 1, -1, -1):
        reversed_arr.append(arr[i])
    return reversed_arr

# Main Program
n = int(input("How many numbers do you want to enter (N)? "))
original_list = []

for i in range(n):
    val = int(input(f"Enter number {i + 1}: "))
    original_list.append(val)

reversed_list = reverse_array_manually(original_list)

print(f"\nOriginal List: {original_list}")
print(f"Reversed List: {reversed_list}")
```

```python
def fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    
    fib_list = [0, 1]
    for i in range(2, n):
        next_fib = fib_list[i - 1] + fib_list[i - 2]
        fib_list.append(next_fib)
    return fib_list

# Main Program
terms = int(input("Enter the number of Fibonacci terms (n): "))
result = fibonacci(terms)
print(f"First {terms} Fibonacci numbers: {result}")
```

```python
def is_prime(n):
    if n <= 1:
        return False
    # Check factors from 2 up to the square root of n
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

# Main Program
print("Prime numbers between 1 and 100:")
primes = []
for number in range(1, 101):
    if is_prime(number):
        primes.append(number)

print(primes)
```

```python
def assign_grade(percentage):
    if percentage >= 90:
        return "A"
    elif percentage >= 75:
        return "B"
    elif percentage >= 60:
        return "C"
    else:
        return "D"

# Main Program
marks = []
print("Enter marks for 5 subjects (out of 100):")
for i in range(5):
    score = float(input(f"Subject {i + 1}: "))
    marks.append(score)

total_marks = sum(marks)
percentage = (total_marks / 500) * 100
grade = assign_grade(percentage)

print(f"\nMarks: {marks}")
print(f"Total Marks: {total_marks:.2f} / 500")
print(f"Percentage:  {percentage:.2f}%")
print(f"Grade:       {grade}")
```

```python
import time

def timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        execution_time = end_time - start_time
        print(f"Execution time for '{func.__name__}': {execution_time:.6f} seconds")
        return result
    return wrapper

@timer
def find_primes_up_to(limit):
    primes = []
    for num in range(2, limit + 1):
        is_prime = True
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(num)
    return primes

# Main Program
prime_list = find_primes_up_to(10000)
print(f"Total primes found up to 10,000: {len(prime_list)}")
```

```python
def validate(func):
    def wrapper(*args, **kwargs):
        # Check positional arguments
        for arg in args:
            if not isinstance(arg, (int, float)) or arg <= 0:
                print(f"Validation Error: All arguments must be positive numbers. Invalid input: {arg}")
                return None
        # Check keyword arguments if any are passed
        for arg in kwargs.values():
            if not isinstance(arg, (int, float)) or arg <= 0:
                print(f"Validation Error: All arguments must be positive numbers. Invalid input: {arg}")
                return None
        return func(*args, **kwargs)
    return wrapper

@validate
def rectangle_area(length, width):
    area = length * width
    print(f"Area of rectangle ({length} x {width}): {area}")
    return area

# Main Program
rectangle_area(10, 5)    # Valid call
rectangle_area(-4, 5)    # Triggers error message
rectangle_area(8, 0)     # Triggers error message
```

```python
def find_duplicates(lst):
    seen = []
    duplicates = []
    
    for item in lst:
        if item in seen:
            # Add to duplicates only once
            if item not in duplicates:
                duplicates.append(item)
        else:
            seen.append(item)
            
    return duplicates

# Main Program
sample_list = [4, 2, 7, 4, 8, 2, 9, 1, 4, 7]
duplicate_items = find_duplicates(sample_list)

print(f"Original list:   {sample_list}")
print(f"Duplicate items: {duplicate_items}")
```

```python
def add_matrices(a, b):
    result = []
    for i in range(len(a)):
        row = []
        for j in range(len(a[0])):
            row.append(a[i][j] + b[i][j])
        result.append(row)
    return result

def input_matrix(name):
    print(f"Enter elements for {name} (3x3):")
    matrix = []
    for i in range(3):
        row = []
        for j in range(3):
            val = int(input(f"  Element [{i + 1}][{j + 1}]: "))
            row.append(val)
        matrix.append(row)
    return matrix

def print_matrix(matrix, title):
    print(f"\n{title}:")
    for row in matrix:
        print("  " + "  ".join(f"{val:4d}" for val in row))

# Main Program
matrix_a = input_matrix("Matrix A")
matrix_b = input_matrix("Matrix B")

sum_matrix = add_matrices(matrix_a, matrix_b)

print_matrix(matrix_a, "Matrix A")
print_matrix(matrix_b, "Matrix B")
print_matrix(sum_matrix, "Sum Matrix (A + B)")
```

```python
def word_frequency(sentence):
    # Convert to lowercase and strip basic punctuation for accurate counting
    cleaned_sentence = ""
    for char in sentence.lower():
        if char.isalnum() or char.isspace():
            cleaned_sentence += char
            
    words = cleaned_sentence.split()
    freq_dict = {}
    
    for word in words:
        if word in freq_dict:
            freq_dict[word] += 1
        else:
            freq_dict[word] = 1
            
    return freq_dict

# Main Program
text = input("Enter a sentence: ")
frequencies = word_frequency(text)

# Sort dictionary items by count (index 1 of the tuple) in descending order
sorted_words = sorted(frequencies.items(), key=lambda item: item[1], reverse=True)

print("\nWord Frequencies (Highest to Lowest):")
for word, count in sorted_words:
    print(f"{word}: {count}")
```
